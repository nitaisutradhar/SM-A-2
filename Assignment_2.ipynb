**2.8 : Create two sets of colors and find their union and intersection.**

```python

set_a = {"red", "green", "blue", "yellow"}
set_b = {"blue", "purple", "red", "orange"}

# Find the union (all unique elements in both sets)
union_set = set_a | set_b

# Find the intersection (elements common to both sets)
intersection_set = set_a & set_b

print(f"Set A: {set_a}")
print(f"Set B: {set_b}")
print(f"Union (A | B): {union_set}")
print(f"Intersection (A & B): {intersection_set}")
```

**2.9 : Write a program that checks if a number is positive, negative, or zero.**

```python

# Function to check if a number is positive, negative, or zero
def check_number(number):
    """
    Takes a number and determines if it is positive, negative, or zero.
    """
    # Check 1: If the number is greater than 0, it is positive.
    if number > 0:
        return f"{number} is a positive number."

    # Check 2: check if it is equal to 0.
    elif number == 0:
        return f"{number} is zero."

    # Check 3: If it is neither positive nor zero, it must be negative.
    else:
        return f"{number} is a negative number."


# Test Case 1: Positive number
result_positive = check_number(15)
print(f"Test Case 1: {result_positive}")

# Test Case 2: Zero
result_zero = check_number(0)
print(f"Test Case 2: {result_zero}")

# Test Case 3: Negative number
result_negative = check_number(-7.5)
print(f"Test Case 3: {result_negative}")
```

**2.10 : Write a program that checks eligibility for a discount (student AND under 25, OR senior over 60).**

```python
def check_discount_eligibility(age):

  # Rule 1: Student AND under 25
  student_discount = age < 25

  # Rule 2: Senior over 60
  senior_discount = age > 60

  # Combine rules using OR logic
  is_eligible = student_discount or senior_discount

  if is_eligible:
    return f"Age: {age}, Status: Eligible for discount."
  else:
    return f"Age: {age}, Status: Not eligible for discount."

# Testing the function with various cases

# Case 1
print(check_discount_eligibility(20))

#case 2
print(check_discount_eligibility(65))

#case 3
print(check_discount_eligibility(30))
```
